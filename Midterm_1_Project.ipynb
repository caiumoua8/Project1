```python
import re
import string

def remove_stopwords(text):

    stopwords = {
        "a", "an", "the", "and", "or", "but",
        "i", "you", "he", "she", "we", "they",
        "me", "my", "your", "our", "their", "it",
        "is", "am", "are", "was", "were", "be", "been", "being",
        "do", "does", "did", "have", "has", "had",
        "in", "on", "at", "to", "for", "of", "with", "from", "by",
        "yeah", "oh", "ooh", "la", "na"
    }
    text = text.lower()

    clean_text = ""
    for char in text:
        if char not in string.punctuation:
            clean_text += char

    words = clean_text.split()

    filtered_words = []
    for word in words:
      if word not in stopwords:
        filtered_words.append(word)

    return filtered_words
```

```python
cleansong1 = remove_stopwords(song1)
cleansong2 = remove_stopwords(song2)
cleansong1
```

```python
def splitcount_lyrics(lyrics):

    counts = {}
    for w in lyrics:
        if w in counts:
            counts[w] += 1
        else:
            counts[w] = 1

    return counts
```

```python
splitsong1 = splitcount_lyrics(cleansong1)
splitsong2 = splitcount_lyrics(cleansong2)
splitsong1
```

```python
def filter_counts(counts):
    items = list(counts.items())

    def get_count(pair):
        return pair[1]

    items.sort(key=get_count, reverse=True)

    top5 = items[:5]
    return top5

def one_counts(counts):
    ones = []
    for word, freq in counts.items():
        if freq == 1:
            ones.append((word, freq))
    return ones
```
